```python
dataset_missed["calls_count"].mean()
```

```python
interested_value = 275

alpha = 0.05

results = st.ttest_1samp(dataset_missed["calls_count"],interested_value)

print('valor p:', results.pvalue / 2)

if (results.pvalue / 2 < alpha) and (dataset_missed["calls_count"].mean() < interested_value):
    print(
        "Rechazamos la hipótesis nula: se tiene una cantidad de llamadas entrantes perdidas inferior a la media "
    )
else:
    print(
        "No podemos rechazar la hipótesis nula: las llamadas entrantes no son significativamente inferiores a la media"
    )
```

```python
#Procederemos a realizar un test A/B para visualizar si existen diferencias entre las 3 tarifas con las hipotesis puestas a prueba anteriormente, solamente que ahora las agruparemos por "user_id".
```

```python
merged_user.head()
```

```python
merged_user["tariff_plan"].value_counts()
```

```python
is_missed_call = [1]
tariff=merged_user[merged_user['is_missed_call'].isin(is_missed_call)][["user_id","calls_count","tariff_plan"]]
```

```python
filtroa = ["A"]
tariff_a=tariff[tariff["tariff_plan"].isin(filtroa)][["user_id","calls_count"]]
```

```python
tariff_a=tariff_a.groupby(["user_id"])["calls_count"].sum().reset_index()
```

```python
tariff_a
```

```python
filtrob = ["B"]
tariff_b=tariff[tariff["tariff_plan"].isin(filtrob)][["user_id","calls_count"]]
```

```python
tariff_b=tariff_b.groupby(["user_id"])["calls_count"].sum().reset_index()
```

```python
tariff_b
```

```python
filtroc = ["C"]
tariff_c=tariff[tariff["tariff_plan"].isin(filtroc)][["user_id","calls_count"]]
```

```python
tariff_c=tariff_c.groupby(["user_id"])["calls_count"].sum().reset_index()
```

```python
tariff_c
```

```python
sample_a=tariff_a["calls_count"]
sample_b=tariff_b["calls_count"]
sample_c=tariff_c["calls_count"]
```

```python
alpha = 0.05

results_AB = st.ttest_ind(
    sample_a,
    sample_b)

results_BC = st.ttest_ind(
    sample_b,
    sample_c)

results_AC = st.ttest_ind(
    sample_a,
    sample_c)

bonferroni_alpha = alpha / 3

print('valor p para comparar los grupos A y B: ', results_AB.pvalue)
print('valor p para comparar los grupos B y C: ', results_BC.pvalue)
print('valor p para comparar los grupos A y C: ', results_AC.pvalue)

if (results_AB.pvalue < bonferroni_alpha):
    print("Hipótesis nula rechazada para los grupos А y B")
else:
    print("Hipótesis nula no rechazada para los grupos А y B")

if (results_BC.pvalue < bonferroni_alpha):
    print("Hipótesis nula rechazada para los grupos B y C")
else:
    print("Hipótesis nula no rechazada para los grupos B y C")

if (results_AC.pvalue < bonferroni_alpha):
    print("Hipótesis nula rechazada para los grupos A y C")
else:
    print("Hipótesis nula no rechazada para los grupos А y C")
```
